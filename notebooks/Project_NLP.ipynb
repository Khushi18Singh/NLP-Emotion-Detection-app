# TEXT Cleaning

### Covert all data of text column in lowercase:-


```python
df['text'] = df['text'].apply(lambda x : x.lower())
```

### Step 2: Removing Punctuations

```python
import string 

def remove_pucn(txt):
    return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_pucn)
```

### Step 3: Removing Numbers

```python
def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new 
df['text'] = df['text'].apply(remove_numbers)
```

### Step 4: Remove Emojies 

```python
def remove_emojis(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new = new + i
    return new

df['text'] = df['text'].apply(remove_emojis)
```

```python
def remove(txt):
  words = txt.split()
  cleaned = []
  for i in words:
    if not i in stop_words:
      cleaned.append(i)

  return ' '.join(cleaned)
     
```

```python
df['text'] = df['text'].apply(remove)
```

# Example of CountVectorizer 

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "I love Pizza",
    "Pizza is the best",
    "I love pasta",
    "Pasta is great"
]

vectorsize = CountVectorizer(ngram_range=(1,3))

X = vectorsize.fit_transform(documents)

print("Vocabulary:", vectorsize.get_feature_names_out())
print("\nBoW Matrix:\n", X.toarray())
```

# Example of TF-IDF Vector

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "I love Pizza",
    "Pizza is the best",
    "I love pasta",
    "Pasta is great"
]

vectorsize = TfidfVectorizer()

X = vectorsize.fit_transform(documents)

print("Vocabulary:", vectorsize.get_feature_names_out())
print("\nBoW Matrix:\n", X.toarray())
```

# Bag of words (Countvectorizer) with NAIVE BAYES Model

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['text'], df['emotion'],
test_size=0.20, random_state=42)
     
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
bow_vectorizer = CountVectorizer()
```

```python
X_train_bow = bow_vectorizer.fit_transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
# MODEL USED NAIVE BAYES
nb_model = MultinomialNB()
```

```python
nb_model.fit(X_train_bow, y_train)
```

#### PREDICTIONS 

```python
pred_bow = nb_model.predict(X_test_bow)
```

```python
pred_bow
```

#### ACCURACY

```python
print(accuracy_score(y_test, pred_bow))
     
```

# TF-IDF Vectorizer With NAIVE BAYES Model

```python
tfidf_vectorizer = TfidfVectorizer()
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)
```

```python
nb2_model = MultinomialNB()
nb2_model.fit(X_train_tfidf,y_train)
```

#### PREDECTIONS 

```python
y_pred = nb2_model.predict(X_test_tfidf)
```

```python
y_pred
```

#### ACCURACY

```python
print(accuracy_score(y_test, y_pred))
```
